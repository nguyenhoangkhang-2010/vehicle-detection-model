```python
import torch
import cv2
import matplotlib.pyplot as plt

from ultralytics import YOLO

model_untrained = YOLO('yolo26n.pt') 
model_trained = YOLO('outputs/vehicle_detection_v1/weights/best.pt')

img_path = 'test.jpg'
results_u = model_untrained(img_path)
results_t = model_trained(img_path)

img_u = results_u[0].plot()
img_t = results_t[0].plot()

fig, axes = plt.subplots(1, 2, figsize=(20, 10))

axes[0].imshow(cv2.cvtColor(img_u, cv2.COLOR_BGR2RGB))
axes[0].set_title('Untrained Model (Baseline)')
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(img_t, cv2.COLOR_BGR2RGB))
axes[1].set_title('Trained Model (Custom)')
axes[1].axis('off')

plt.show()
```
