```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('../outputs/vehicle_detection_v4/results.csv')
data.columns = data.columns.str.strip()

plt.figure(figsize=(10, 5))
plt.plot(data['epoch'], data['metrics/mAP50(B)'], label='mAP@0.5', color='blue', linewidth=2)
plt.title('Sự tăng trưởng mAP@0.5 qua 100 Epochs')
plt.xlabel('Epoch')
plt.ylabel('mAP@0.5')
plt.grid(True)
plt.legend()
plt.savefig('map_plot.png', dpi=300)
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(data['epoch'], data['train/box_loss'], label='Box Loss')
plt.plot(data['epoch'], data['train/cls_loss'], label='Class Loss')
plt.title('Sự suy giảm Loss trong quá trình huấn luyện')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True)
plt.legend()
plt.savefig('loss_plot.png', dpi=300)
plt.show()
```
